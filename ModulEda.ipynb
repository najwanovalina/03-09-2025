```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
```

Tugas 2:
Perhitungan statistik (seperti rata-rata total penjualan atau pendapatan) menjadi tidak akurat atau menghasilkan nilai NaN. Serta Proses pembuatan laporan/grafik visualisasi data dapat mengalami error atau menampilkan data yang menyesatkan

```python
import pandas as pd

df = pd.read_excel('dataset_nilai_akademik_siswa.xlsx')
print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

Tugas 3:
Jumlah baris total = 79 baris.Kolom nilai dan guru_pengampu hanya memiliki 75 nilai non-null.   Artinya: Terdapat 4 baris data yang kosong (missing values) pada kolom nilai dan guru_pengampu

```python
# Cek data kosong
print(df.isnull().sum())

# Penanganan missing value
df['nilai'] = df['nilai'].fillna(df['nilai'].median()) # atau nilai kriteria rata-rata
df['guru_pengampu'] = df['guru_pengampu'].fillna('Belum Terdata')
```

Tugas 4:
Kolom angka (seperti terjual atau nilai) dapat diisi dengan nilai perkiraan rasional (imputation) seperti nilai 0, rata-rata (mean), atau nilai tengah (median) agar jumlah sampel data tidak berkurang.   Kolom kategori utama (seperti menu atau nama siswa) yang hilang sulit ditebak secara pasti. Jika nama menu/siswa kosong, baris tersebut kehilangan identitas/maknanya sehingga lebih aman dihapus (dropna)

```python
# Cek dan hapus duplikat
print("Jumlah duplikat:", df.duplicated().sum())
df = df.drop_duplicates()

# Bersihkan format teks pada kolom nilai (misal '67 poin') dan ubah tipe data
df['nilai'] = df['nilai'].astype(str).str.replace(' poin', '')
df['nilai'] = pd.to_numeric(df['nilai'], errors='coerce')

# Hapus data terdistorsi/outlier ekstrem (seperti nilai 999) jika ada
df.loc[df['nilai'] > 100, 'nilai'] = None
df['nilai'] = df['nilai'].fillna(df['nilai'].median())

df['nilai'] = df['nilai'].astype(int)
print(df.dtypes)
```

Jumlah baris setelah drop_duplicates(): Dari 79 baris, berkurang menjadi 75 baris (terdapat 4 baris duplikat yang dihapus).   Pentingnya memastikan tipe data (dtypes) sudah benar: Jika kolom numerik dianggap bertipe object (teks/string), kita tidak bisa melakukan operasi matematika (seperti .mean(), .sum(), atau pemfilteran perbandingan angka seperti nilai > 75)

```python
# Filtering (Siswa dengan nilai tuntas >= 75)
tuntas = df[df['nilai'] >= 75]

#Diurutkan berdasarkan nilai tertinggi
urut = df.sort_values(by='nilai', ascending=False)

# Kolom turunan
df['status_kelulusan'] = df['nilai'].apply(lambda x: 'Lulus' if x >= 75 else 'Tidak Lulus')

#Rata-rata nilai per mata pelajaran
ringkasan = df.groupby('mata_pelajaran')['nilai'].mean().reset_index()
print(ringkasan)
```
