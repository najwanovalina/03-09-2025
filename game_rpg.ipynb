Mengirimkan objek utuh sebagai parameter sangat penting karena method serang perlu memanipulasi data dan memanggil fungsi milik objek tersebut secara langsung misalnya memanggil lawan.diserang.untuk mengurangi nilai lawan.hp, Jika yang dikirim hanya berupa string nama seperti "Zilong", program hanya mengetahui namanya saja tetapi tidak memiliki akses ke status/data internal hero tersebut seperti sisa HP, sehingga proses pertarungan dan pengurangan damage tidak dapat dijalankan secara nyata pada objek yang bersangkutan


```python
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)
eudora.skill_fireball(balmond)
```

 Pertanyaan 2:
 Fungsi super() memiliki fungsi memangggil method atau constructor (__init__) dari Parent class (Induk) ke dalam child class (Anak). Perannya adalah meneruskan parameter awal dari class Anak agar diolah dan didaftarkan oleh logika yang ada di class Induk. Hal ini memungkinkan class Anak memanfaatkan atau memperluas atribut dan perilaku Induk tanpa harus menulis ulang kodenya dari awal (code reusability)

```python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        self.__hp = hp_awal
    def get_hp(self):
        return self.__hp

    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

hero1 = Hero("Layla", 100)
hero1.set_hp(-50)
print(hero1.get_hp())
print(f"Mencoba akses paksa: {hero1._Hero__hp}")
```

```python
from abc import ABC, abstractmethod
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass
class Hero(GameUnit):

    def __init__(self, nama):
        self.nama = nama
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")
    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")
    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

 Alasan kenapa Harus Sama karena Polimorfisme mengandalkan keseragaman nama method. Struktur perulangan hanya memanggil method bernama serang(), sehingga jika nama method di class anak berbeda, fungsi khusus milik class anak tersebut tidak akan tereksekusi


```python
from abc import ABC, abstractmethod
class KamarHotel(ABC):
    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        # Encapsulation: Atribut Private
        self.__stok = 0
        self.__harga_dasar = harga_dasar
        self.set_stok(stok)

    def get_stok(self):
        return self.__stok

    def set_stok(self, jumlah):
        if jumlah < 0:
            print(f"Gagal update stok {self.nama_kamar}! Stok tidak boleh negatif ({jumlah}).")
        else:
            self.__stok = jumlah
            print(f"Berhasil menambahkan stok {self.nama_kamar}: {self.__stok} unit.")

    def get_harga_dasar(self):
        return self.__harga_dasar


    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass

class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    def hitung_harga_total(self, jumlah_malam):
        pajak_per_malam = self.get_harga_dasar() * 0.10
        total_per_malam = self.get_harga_dasar() + pajak_per_malam
        return total_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam):
        pajak = int(self.get_harga_dasar() * 0.10)
        subtotal = int(self.hitung_harga_total(jumlah_malam))
        print(f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}")
        print(f"   Harga Dasar/Malam: Rp {self.get_harga_dasar():,} | Pajak(10%): Rp {pajak:,}")
        print(f"   Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,}".replace(',', '.'))


class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    def hitung_harga_total(self, jumlah_malam):
        pajak_per_malam = self.get_harga_dasar() * 0.05
        total_per_malam = self.get_harga_dasar() + pajak_per_malam
        return total_per_malam * jumlah_malam

    def tampilkan_detail(self, jumlah_malam):
        pajak = int(self.get_harga_dasar() * 0.05)
        subtotal = int(self.hitung_harga_total(jumlah_malam))
        print(f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}")
        print(f"   Harga Dasar/Malam: Rp {self.get_harga_dasar():,} | Pajak(5%): Rp {pajak:,}")
        print(f"   Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,}".replace(',', '.'))

def proses_transaksi(daftar_pesanan):
    print("--- STRUK PEMESANAN ---")
    total_tagihan = 0

    for idx, (kamar, malam) in enumerate(daftar_pesanan, start=1):
        print(f"{idx}. ", end="")
        kamar.tampilkan_detail(malam)
        total_tagihan += kamar.hitung_harga_total(malam)

    print("-" * 40)
    print(f"TOTAL TAGIHAN: Rp {int(total_tagihan):,}".replace(',', '.'))
    print("-" * 40)

if __name__ == "__main__":
    print("--- SETUP DATA KAMAR ---")
    deluxe1 = KamarDeluxe("Kamar Deluxe Sea View", 10, 1500000, "Private Pool")

    standard1 = KamarStandard("Kamar Standard Superior", -5, 500000, "2 Orang")

    standard1.set_stok(20)
    print()
    pesanan_tamu = [
        (deluxe1, 2),
        (standard1, 1)
    ]

    proses_transaksi(pesanan_tamu)
```
